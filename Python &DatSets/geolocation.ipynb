```python
def calculate_distance(lat_seller , lat_customer , lon_seller , lon_customer):
    lat_seller, lat_customer , lon_seller , lon_customer = map(np.radians , [lat_seller , lat_customer ,lon_seller ,lon_customer])
    dlat = lat_customer - lat_seller
    dlon = lon_customer - lon_seller
    a = np.sin(dlat/2)**2 + np.cos(lat_customer)* np.cos(lat_seller) *np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    # نصف قطر الأرض بالكيلومتر ~ 6371 كم
    km = 6371 * c
    return km
```

```python
order_items['distance_km'] = order_items.apply(lambda raw: calculate_distance(
raw['sellers_lat'],
raw['customer_lat'],
raw['sellers_lon'],
raw['customer_lon']    
),axis=1)
```
